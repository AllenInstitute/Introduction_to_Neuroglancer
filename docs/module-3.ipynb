## Overview

In this error correction module, students learn how to identify and fix one of the most common errors in automated neuron reconstructions: the **false merger**, where two or more separate cells (or fragments of cells) have been incorrectly joined into a single segment. Here we will learn how to fix these errors by navigating the 2D electron microscopy (EM) images and 3D meshes in Neuroglancer, place annotations, use the Find Path tool to trace the connection between two points, and use the Split tool to separate falsely merged segments.

## Learning Goals 

* Navigate fluidly between the 2D EM cross-sections and the 3D mesh view in Neuroglancer
* Place and manage point annotations to mark locations of interest
* Use the Find Path tool to locate where a false merger occurs
* Use the Split tool to separate falsely merged segments
* Use the mesh bounding box to focus on a local region of a large segment


## Background and Introduction 

Many connectome datasets are reconstructed automatically by machine learning algorithms that trace each cell through thousands of serial EM images. While these algorithms are remarkably accurate, they are not perfect. Two kinds of errors are common:

* **False mergers** – two separate objects (e.g., an axon from one cell and a dendrite from another) are incorrectly combined into a single segment.
* **False splits** – a single object is incorrectly broken into two or more separate segments.

The process of finding and fixing these errors by hand is called **proofreading**. In this module we focus on **false mergers** and how to correct them by **splitting**.

<!-- TODO: replace with an example figure of a false merger (e.g., an axon crossing a dendrite) -->
![Figure 1. Example of a false merger. Two different neurons have been joined where their processes come into close contact.](resources/module-2/false_merger_example.png)

In these datasets, every segment is built from many small pieces called **supervoxels**. When you perform a split, you are telling the proofreading system (CAVE) which supervoxels belong on each side of the cut. The system then calculates the best place to separate the segment for you.

### 3. Annotations

Annotations have many use cases and are often used to keep track of your work, or to share information between proofreaders. For example, you might place a point on each location where you suspect a false merger so that you can come back to it later.

To place a point annotation:

1. Select (or create) an annotation layer by {{< kbd Right-Click >}} on the layer tab. If you do not have an annotations tab, {{< kbd Right-Click >}} on the + in the layer tabs.
2. In the **Annotations** tab of the right-hand panel, click the point tool.
3. {{< kbd Ctrl-Left-Click >}} in the 2D or 3D view to place a point.

Each annotation will appear in the list in the right-hand panel. Clicking on an annotation in the list will move the view to that location. You can also add a description to an annotation to make notes for yourself or other proofreaders.

### 4. Find Path Tool

The **Find Path** function allows you to locate a connective path between any two points placed along your cell. This can be very useful when trying to navigate through a large cell, or to locate the origin point of a merger. For example, if an object has two cell bodies you can place on point on each cell body. In examples that are a less obvious, you cam place one point on the cell body of your neuron and a second point on a branch that you think belongs to a different cell. In either case, the path will show you exactly where the two are connected.

To use Find Path:

<!-- TODO: verify keybinding / tool location for Find Path in Spelunker -->
1. Select the segmentation layer and open the **Graph** tab in the right-hand panel.
2. Choose the **Find Path** tool.
3. {{< kbd Ctrl-Left-Click >}} to place a **source** point on the segment.
4. {{< kbd Ctrl-Left-Click >}} to place a **target** point on the same segment.
5. Click **Find Path**. A line will be drawn along the shortest path through the segment between the two points.

<!-- TODO: add image (source: blog.pyr.ai Screen-Shot-2022-04-27-at-11.05.25-AM-1024x822.png) -->
![Figure 5. A path found between two points on a cell.](resources/module-2/find_path.png)

Follow the path in the 3D view and look for the point where it passes from one cell into another. Then, inspect this location in the 2D EM images to confirm that it is a false merger.

:::{.note}
In the case that there are multiple merge points between the two objects you are splitting, each time you make a split the Find Path tool will refresh to point you towards where the next false merge may be.
:::


### 5. Splitting Tool

Once you have found the false merge point, it is time to split the segments. To perform a split:

1. Press {{< kbd C >}} to engage split mode (think C for "cut"). Similar to the find path tool, you should see a little tool bar show up in the lower left corner.
2. **Red** split points are automatically engaged. {{< kbd Ctrl-Left-Click >}} to place them on the part of the segment you want to **keep on one side** of the split.
3. Press {{< kbd G >}} to toggle to **blue** split points. {{< kbd Ctrl-Left-Click >}} to place them on the part of the segment you want on the **other side** of the split.
4. Press {{< kbd Enter >}} to execute the split!

:::{.callout-tip}
Placing more than one point on each side can help the split algorithm find the right place to cut, especially when the merger is between two large objects.
:::

After the split is complete, the red/blue annotation points will disappear and the segment will be given a new Segment ID. If that was the final merge point, you will see two objects on either side of the split, each with a unique Segment ID and a different color. If there are any remaining merge points, the object will have a different Segment ID than before but will still be one object (same color). Once you have successfully separated the object apart, check both segments in 3D to make sure the split happened where you expected.

:::{.callout-caution}

Splits made in a proofreading dataset are saved and shared with every other user so please double check your split points before pressing {{< kbd Enter >}}.
:::

### 6. Voxel-Perfect Isn't Required

A common worry when splitting is that you must place your points precisely at the boundary between the two cells. **This is not the case.** Because the split algorithm works on the graph of supervoxels, it only needs to know which pieces belong on each side. It then computes the minimal cut between your red and blue points on its own.

This means:

* Your points only need to be clearly on the correct side of the merger – they do not need to be right at the junction.
* Points placed a short distance away from the merger, on clearly identifiable parts of each cell, often give better results than points squeezed up against the junction.
* If a split does not come out as expected, you can refine it by adding more points and trying again.
* Because voxels are organized into supervoxels (think of them as clusters of voxels), if the split has a few voxels that cross your intended split boundary, that is ok. No need to refine it since you cannot split a supervoxel.

### 7. Mesh Bounding Box

Neurons that are falsely merged with many other objects can become very large, making the 3D view slow and cluttered. To focus on the area you are working on, you can limit the meshes so that only the portion inside a bounding box around your current location is displayed.

<!-- TODO: verify the name and location of the mesh bounding box setting in Spelunker -->
1. Select the segmentation layer and open the **Render** tab in the right-hand panel.
2. Enable the option to restrict meshes to a bounding box and set its size.
3. Move around the volume – only the parts of the meshes inside the box around the view center will be shown.

Remember to turn this option off when you want to view the entire neuron again.


## Helpful Tips / FAQ

**My segment won't load or shows a red error message.**
Make sure you have accepted the Terms of Service (<https://global.daf-apis.com/sticky_auth/api/v1/tos/2/accept>) and then hard refresh your Neuroglancer window. If the problem persists, call your instructor/TA.

**I'm stuck in split mode (or another mode).**
Look at the footer at the bottom of the screen to see which mode is active, and press the same key again (e.g., {{< kbd C >}}) to exit.

**My split didn't cut where I expected.**
You cannot undo the split but you can add more red and blue points on clearly identifiable parts of each cell. Remember that the points do not need to be voxel-perfect.

**The 3D view is slow or cluttered.**
Hide segments you aren't working on, or use the mesh bounding box to show only the region around your current location.

**My Segment ID changed after I made an edit but it is still one object.**
This can happen! If the Segmend ID has changed that means that the split you attempted to make has occured but it is likely that these two objects are merged elsewhere. Use the Find Path tool to find where the next merge might be. Note that if you were using the Mesh Bounding Box you will want to broaden the few since the next merge may be outside the current view.
